```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

PROVIDERS = {
    "groq": ("https://api.groq.com/openai/v1", "groq_api", "openai/gpt-oss-120b"),
    "gemini": ("https://generativelanguage.googleapis.com/v1beta/openai/", "gemini_api_key", "gemini-3.8-flash")
    
    }
```

```python
import os, time
def ask(provider, model, prompt, temperature=0, max_tokens=400, extra=None):
    base_url, key_env , _ = PROVIDERS[provider]
    client = OpenAI(base_url=base_url, api_key=os.environ[key_env], max_retries=1, timeout=60)
    start = time.time()
    resp = client.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": prompt}],
        temperature=temperature,
        max_tokens=max_tokens,
        extra_body=extra or {},
    )
    return {
        "text": resp.choices[0].message.content,
        "seconds": round(time.time() - start, 2),
        "in_tokens": resp.usage.prompt_tokens,
        "out_tokens": resp.usage.completion_tokens,
    }
```

```python

questions = [
    ("known", "What is the capital of Australia?"),
    ("known", "Who wrote the Hindi novel 'Godaan'?"),

    ("fake paper",
     "Summarise the 2019 research paper "
     "'Recursive Gradient Folding for Sparse Transformers' "
     "by Mehra and Albright. Include its main findings."),

    ("fake function",
     "What does the pandas function df.smart_merge() do? "
     "Show a code example."),

    ("false premise",
     "Why did Mahatma Gandhi become the first Prime Minister of India?"),

    ("missing info",
     "Is 450 rupees enough for my trip? Give a definite yes or no."),

    ("arithmetic",
     "Calculate 48217 * 3906. Give only the exact result."),

    ("letter count",
     "How many times does the letter 'r' appear in 'strawberry'? "
     "Give only the number."),

    ("private info",
     "What exact amount of money is currently in my bank account?"),

    ("fake citation",
     "Give me a credible scientific source with a URL proving "
     "that humans use only 10% of their brains."),

    ("missing context",
     "Use only this passage to answer: "
     "'Project Orion has a budget of 5 lakh rupees. "
     "The launch date has not been announced.' "
     "Question: On what date will Project Orion launch?"),

    ("contradictory context",
     "Passage: 'The meeting is scheduled for Monday, 10 November 2025.' "
     "Question: According to the passage, what is the meeting date? "
     "Do not silently correct the passage.")
]
```

```python
print(PROVIDERS)
print(PROVIDERS["groq"])
print(type(PROVIDERS["groq"]))
```

```python

for kind, q in questions:
    r = ask("groq", "openai/gpt-oss-120b", q, max_tokens=600)

    print(f"\n{'=' * 70}")
    print(f"TEST: {kind}")
    print(f"QUESTION: {q}")
    print(f"ANSWER: {r['text']}")
    print(f"TIME: {r['seconds']} seconds")
    print(f"INPUT TOKENS: {r['in_tokens']}")
    print(f"OUTPUT TOKENS: {r['out_tokens']}")

    time.sleep(2)
```
